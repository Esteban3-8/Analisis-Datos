```python
import pandas as pd

# Definir la ruta del archivo CSV
ruta_archivo = 'data/ventas.csv'  # Ajusta la ruta si es necesario

# Cargar el archivo CSV en un DataFrame llamado df
df = pd.read_csv(ruta_archivo)

df.describe()
```

```python
# Número total de registros (filas)
total_registros = df.shape[0]
print("Número total de registros:", total_registros)
```

```python
# Columnas que no son numéricas
columnas_no_numericas = df.select_dtypes(exclude=['int64','float64']).columns
print("Columnas no numéricas:", list(columnas_no_numericas))
```

```python
valores_nulos = df.isna().sum()
print("Cantidad de valores nulos por columna:")
print(valores_nulos)
```

```python
# Contar registros duplicados en el DataFrame
registros_duplicados = df.duplicated().sum()

print("Número de registros duplicados en el DataFrame:", registros_duplicados)
```

```python
# 1. Rellenar valores nulos en la columna 'descuento' con 0
df['descuento'] = df['descuento'].fillna(0)

# 2. Eliminar filas donde 'region' o 'vendedor' sean nulas
df = df.dropna(subset=['region', 'vendedor'])

# Verificar resultados
print("Cantidad de valores nulos por columna después del tratamiento:")
print(df.isna().sum())
```

```python
# Eliminar registros duplicados
df = df.drop_duplicates()

# Verificar la nueva forma del DataFrame
print("Dimensiones del DataFrame después de eliminar duplicados:", df.shape)
# Filtrar las ventas realizadas en la región 'Norte'
df = pd.read_csv(ruta_archivo)

filtra_venta = df[df['region'] == 'Norte']

# Mostrar las primeras filas del resultado
filtra_venta.head()
```

```python

# Filtrar las ventas realizadas en la región 'Norte'
filtra_ventas = df[df['region'] == 'Norte']

# Mostrar las primeras filas del resultado
filtra_ventas.head()
```

```python
# Filtrar ventas del canal 'Online' con más de 5 unidades
venta_online = df[(df['canal'] == 'Online') & (df['unidades'] > 5)]

# Mostrar las primeras filas del resultado
venta_online.head()
```

```python
# Definir las columnas que queremos en el subconjunto
columnas = ['fecha', 'region', 'vendedor', 'producto', 'unidades', 'ingreso_total']

# Crear el subconjunto del DataFrame
df_SUB = df[columnas]

# Mostrar las primeras filas
df_SUB.head()
```

```python
# Suma de la columna ingreso_total
ingreso_total_periodo = df['ingreso_total'].sum()
print(ingreso_total_periodo)
```

```python
# Promedio de ingreso por registro/venta
ingreso_promedio = df['ingreso_total'].mean()
print(ingreso_promedio)
```

```python
unidades_por_producto = df.groupby('producto')['unidades'].sum()
print("\nTotal de unidades vendidas por producto:")
print(unidades_por_producto)
```

```python
# Agrupar por 'region' y sumar el ingreso total
ingreso_por_region = df.groupby('region')['ingreso_total'].sum().sort_values(ascending=False)

# Mostrar el resultado
print("Ingreso total por región (de mayor a menor):")
print(ingreso_por_region)
```

```python
unidades_por_producto.idxmax()
```

```python

region_top = ingreso_por_region.idxmax()
print("Región que genera el mayor ingreso total:", region_top)
```

```python
tabla_ingreso = pd.pivot_table(df, values='ingreso_total', index='region', columns='canal', aggfunc='sum', fill_value=0)

print("Ingreso total por región y canal:")
print(tabla_ingreso)
```

```python
tabla_promedio_unidades = pd.pivot_table(df, values='unidades', index='categoria', columns='canal', aggfunc='mean', fill_value=0)
print("Promedio de unidades vendidas por categoría y canal:")
print(tabla_promedio_unidades)
```

```python
# Calcular promedio de ingresos para usar como umbral
umbral_bajo = tabla_ingreso.values.mean()

# Filtrar combinaciones región–canal con ingreso_total menor al umbral
bajo_desempeno = tabla_ingreso[tabla_ingreso < umbral_bajo].stack()

# Mostrar combinaciones con bajo desempeño
print("\nCombinaciones región–canal con bajo desempeño (ingreso_total menor al promedio):")
print(bajo_desempeno)
```

```python
# Ingreso total por región
ingreso_region = df.groupby('region')['ingreso_total'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
ingreso_region.plot(kind='bar', color='skyblue')
plt.title('7.1 Ingreso total por región')
plt.xlabel('Región')
plt.ylabel('Ingreso total')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 7.2 Unidades vendidas por producto
unidades_producto = df.groupby('producto')['unidades'].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
unidades_producto.plot(kind='bar', color='salmon')
plt.title('7.2 Unidades vendidas por producto')
plt.xlabel('Producto')
plt.ylabel('Unidades vendidas')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
